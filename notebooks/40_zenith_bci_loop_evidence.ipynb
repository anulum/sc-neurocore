# Zenith BCI Loop Evidence

Evidence Boundary

This notebook validates `sc_neurocore.interfaces.ZenithBCILoop` using synthetic waveform windows only.
It does not claim clinical safety and does not represent human-subject stimulation evidence.
Schema: `sc-neurocore.zenith-bci-loop-evidence.v1`.


```python
from __future__ import annotations

import numpy as np

from sc_neurocore.interfaces import ZenithBCILoop, ZenithBCILoopConfig
```

```python
waveform = np.zeros((96, 4), dtype=np.float32)
waveform[12, 0] = -25.0
waveform[40, 1] = -30.0
waveform[72, 3] = -35.0

loop = ZenithBCILoop(
    ZenithBCILoopConfig(
        n_channels=4,
        sampling_rate_hz=30_000,
        gpu_lanes=4,
        latency_budget_ms=10.0,
        threshold_sigma=4.0,
        snippet_samples=16,
    )
)
result = loop.process_stream(waveform, window_start_us=123, pathway_name="neuropixels")
```

```python
manifest = {
    "schema_version": "sc-neurocore.zenith-bci-loop-evidence.v1",
    "summary": {
        "pathway_name": result.pathway_name,
        "spike_count": result.spike_count,
        "feedback_active_channels": result.feedback_active_channels,
        "command": result.command,
        "latency_budget_ms": result.latency_budget_ms,
        "total_latency_ms": result.total_latency_ms,
        "latency_budget_met": result.latency_budget_met,
        "latency_breakdown_keys": sorted(result.latency_breakdown_ms.keys()),
    },
}
manifest
```
